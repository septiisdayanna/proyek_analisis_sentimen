# Scraping Dataset
Pada tahap ini kita mengambil semua informasi ulasan dari aplikasi di Google Play Store, yaitu Seabank dengan menggunakan ID 'id.co.bankbkemobile.digitalbank'

### Import library

```python
# Mengimpor pustaka google_play_scraper untuk mengakses ulasan dan informasi aplikasi dari Google Play Store.
from google_play_scraper import reviews_all, reviews, Sort 
# Mengimpor pustaka csv untuk menulis data ke file CSV.
import csv
# Mengimpor pustaka pandas untuk manipulasi data.
import pandas as pd 
```

### Scraping semua informasi ulasan dari aplikasi Seabank

```python
# Mengambil 35.000 ulasan dari aplikasi dengan ID 'id.co.bankbkemobile.digitalbank' di Google Play Store.
scrapreview, _ = reviews(                       # Menggunakan fungsi reviews untuk mengambil ulasan
    'id.co.bankbkemobile.digitalbank',          # ID aplikasi
    lang='id',                                  # Bahasa ulasan 
    country='id',                               # Negara 
    sort=Sort.MOST_RELEVANT,                    # Urutan ulasan 
    count=35000                                 # Jumlah maksimum ulasan yang ingin diambil
)
```

### Menyimpan ulasan dalam file CSV

```python
# Mengonversi data ulasan menjadi DataFrame pandas 
app_reviews_df = pd.DataFrame(scrapreview) 
app_reviews_df.to_csv('ulasan_aplikasi.csv', index=False, encoding='utf-8') 
```
